```python
import numpy as np
np.random.seed(0)

X =np.linspace(-1,1,100).reshape(-1,1)
y = 2 * X +1
```

```python
W1 = np.random.randn(1,10)
b1 = np.zeros((1,10))
W2 = np.random.randn(10,1)
b2 = np.zeros((1,1))
```

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x >0).astype(float)
```

```python
z1 = X @ W1 + b1

a1 = relu(z1)

z2 = a1 @ W2 + b2

y_pred = z2 
```

```python
def mse(y,y_hat):
    return np.mean((y -y_hat)**2)

loss = mse(y,y_pred)

loss
```

```python
#backward pass

lr = 0.1

dL_dy = 2* (y_pred - y) / len(y)

dW2 = a1.T @ dL_dy
db2 = np.sum(dL_dy, axis=0, keepdims=True)

da1 = dL_dy @ W2.T

dz1 = da1 * relu_derivative(z1)

dW1 = X.T @ dz1

db1 = np.sum(dz1, axis=0, keepdims=True)

```

```python
#update weights

W1 -= lr * dW1
b1 -= lr * db1
W2 -= lr * dW2
b2 -= lr * db2
```

```python
#training a neural network from scratch


for epoch in range(1000):
    z1 = X @ W1 + b1
    a1 = relu(z1)
    y_pred = a1 @ W2 +b2


    loss = mse(y,y_pred)

    dL_dy = 2 * (y_pred -y) / len(y)

    dW2 = a1.T @ dL_dy
    db2 = np.sum(dL_dy, axis=0, keepdims=True)

    da1 = dL_dy @ W2.T
    dz1 = da1 * relu_derivative(z1)
    dW1 = X.T @ dz1

    db1 = np.sum(dz1,axis=0, keepdims=True)

    W1 -= lr * dW1
    b1 -= lr * db1
    W2 -= lr * dW2
    b2 -= lr * db2


    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.6f}")

```
